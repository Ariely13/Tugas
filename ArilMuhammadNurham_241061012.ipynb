# 0. Mengimport Library yang Dibutuhkan

```python
import numpy as np # Mengimpor library untuk komputasi numerik
import matplotlib.pyplot as plt # Mengimpor library untuk visualisasi data
from scipy.stats import poisson # Mengimpor fungsi statistik khusus Poisson
```

# 1. Menyiapkan Settingan Data

Langkah awal kita menyiapkan parameter dasar untuk distribusi Poisson. Berdasarkan rata-rata 5 pelanggan per 10 menit, maka nilai rata-rata per jam ($\lambda$) ditetapkan sebesar 30. Kita memanggil library numpy untuk mensimulasikan data selama 100 hari ke depan dan scipy.stats untuk kebutuhan perhitungan peluang statistik. Penggunaan fungsi random seed juga diterapkan agar hasil angka acak yang dihasilkan tetap konsisten dan tidak berubah setiap kali kode dijalankan ulang.

```python
lam = 30     # Menetapkan nilai rata-rata 30 pelanggan per jam (5 pelanggan * 6)
hari = 100   # Menetapkan durasi simulasi selama 100 hari
np.random.seed(42) # Menjamin hasil acak tetap konsisten saat dijalankan ulang
data = np.random.poisson(lam, hari) # Menghasilkan data acak berdasarkan distribusi Poisson
```


# 2. Menjalankan Simulasi dan Grafik
Setelah itu kita akan mencoba mengeksekusi simulasi dengan membangkitkan data acak Poisson untuk menggambarkan kedatangan pelanggan selama 100 hari. Hasilnya kemudian divisualisasikan melalui histogram agar pola dan frekuensi kedatangan pelanggan per jam dapat diamati secara jelas menggunakan grafik histogram.


```python
plt.hist(data, bins=15, color='skyblue', edgecolor='black') # Membuat grafik histogram
plt.title('Visualisasi Kedatangan Pelanggan') # Menambahkan judul pada grafik
plt.xlabel('Jumlah Pelanggan per Jam') # Memberikan label pada sumbu X
plt.ylabel('Frekuensi Kejadian (Hari)') # Memberikan label pada sumbu Y
plt.show() # Menampilkan grafik secara utuh
```

 Grafik menunjukkan bahwa jumlah pelanggan per jam mayoritas terpusat pada rentang 25 hingga 35 orang, dengan frekuensi tertinggi berada di sekitar nilai rata-rata 30 pelanggan. Hal ini membuktikan bahwa simulasi data telah berjalan sesuai dengan parameter distribusi Poisson yang ditetapkan, hal ini juga memperlihatkan bahwa jumlah kedatangan yang melebihi angka 40 memiliki frekuensi yang sangat rendah, yang mengindikasikan bahwa risiko kelebihan kapasitas sangat jarang terjadi dalam skenario simulasi 100 hari ini.

 ---


# 3. Menghitung rata-rata & variansi (dari hasil simulasi)

```python
rata_rata = np.mean(data) # Menghitung rata-rata dari data simulasi
variansi = np.var(data) # Menghitung variansi dari data simulasi
print(f"Rata-rata Simulasi: {rata_rata}") # Menampilkan hasil rata-rata ke layar
print(f"Variansi Simulasi: {variansi}") # Menampilkan hasil variansi ke layar
```

## Interpretasi Hasil :
Hasil rata-rata sebesar 30,05 dan variansi sebesar 28,94 menunjukkan angka yang sangat mendekati parameter target (λ=30). Dalam teori distribusi Poisson, nilai rata-rata dan variansi idealnya adalah sama. Kedekatan kedua nilai ini membuktikan bahwa simulasi 100 hari yang dijalankan sudah sangat akurat dan mengikuti karakteristik distribusi Poisson secara konsisten.

# 4. Probabilitas jika pelanggan lebih dari 40
Terakhir menghitung risiko atau peluang terjadinya kondisi di mana jumlah pengunjung melampaui kapasitas maksimal, restoran menggunakan fungsi distribusi kumulatif (CDF) untuk menentukan probabilitas kejadian di luar batas daya tampung normal berdasarkan rata-rata kedatangan yang telah ditetapkan.

```python
prob_lebih = 1 - poisson.cdf(40, lam) # Menghitung probabilitas pelanggan lebih dari 40
print(f"Probabilitas Kelebihan Kapasitas: {prob_lebih:.4f}") # Menampilkan hasil probabilitas
```

diperoleh nilai probabilitas kelebihan kapasitas sebesar 0,0323 atau sekitar 3,23%. Maka peluang terjadinya lonjakan pelanggan yang melampaui kapasitas maksimal 40 orang sangatlah rendah. Dengan demikian, dapat disimpulkan bahwa kapasitas restoran saat ini masih sangat mencukupi untuk melayani rata-rata kedatangan pelanggan tanpa risiko kepadatan yang signifikan.
