```python
# mengimpor library yang diperlukan
import pandas as pd
import numpy as np
```

```python
# baca dataset
df = pd.read_csv('/content/data_tugas 2.csv')

# ganti tanda tanya dengan NaN karena merupakan missing value
df.replace('?', np.nan, inplace=True)

print(df.head())
print(df.shape)
```

Berikut versi yang lebih rapi dan enak dibaca:

Kode tersebut digunakan untuk menginisialisasi dataset dengan membaca file CSV serta melakukan pembersihan data awal. Nilai berupa tanda tanya (`?`) diubah menjadi `NaN` (*Not a Number*) agar dikenali sebagai *missing value* oleh library Pandas dan dapat diproses lebih lanjut. Berdasarkan output `df.shape`, dataset memiliki 205 baris dan 9 kolom.


#1. Menampilkan jumlah missing value

```python
missing = df.isnull().sum()
print("\nJumlah missing value tiap kolom:") # menghitung jumlah nilai kosong (NaN) di setiap kolom
print(missing)

# cari kolom yang paling banyak missing valuenya
print("\nKolom dengan missing value terbanyak:")
print(missing[missing > 0].sort_values(ascending=False))
# menampilkan hanya kolom yang masih memiliki missing value, lalu diurutkan dari yang terbanyak
```

Kode ini digunakan untuk mengidentifikasi sekaligus mengurutkan sebaran nilai yang hilang (*missing values*) dalam dataset. Fungsi `isnull().sum()` menghitung jumlah data kosong pada setiap kolom, kemudian `sort_values()` mengurutkannya dari yang terbanyak hingga paling sedikit. Hasil yang diperoleh menunjukkan bahwa kolom *normalized-losses* memiliki jumlah missing value paling tinggi, yaitu sebanyak 41 data, sehingga perlu menjadi prioritas utama dalam proses pembersihan data agar tidak memengaruhi hasil analisis.


# 2. Ganti missing value pada variabel kategori

```python
# kolom kategori yang ada missing valuenya
kolom_kategori = ['make', 'bahan-bakar', 'jumlah-pintu', 'body-style']

# isi missing value dengan 'unknown'
for col in kolom_kategori: # looping untuk setiap kolom kategori
    df[col].fillna('unknown', inplace=True) # mengisi nilai kosong dengan 'unknown'

# cek hasilnya
print("\nRingkasan kategori setelah diisi unknown:")  # menampilkan judul output
for col in kolom_kategori:                            # looping lagi untuk melihat hasil
    print(f"\n{col}:")                                # menampilkan nama kolom
    print(df[col].value_counts())                     # menampilkan jumlah tiap kategori
```

Kode tersebut digunakan untuk menangani missing value pada kolom kategori dalam dataset. Sebuah list dibuat untuk menentukan kolom yang akan diproses, kemudian dilakukan perulangan untuk mengisi nilai kosong (NaN) dengan `'unknown'` menggunakan fungsi `fillna()` agar data lebih konsisten. Setelah itu, ditampilkan ringkasan setiap kolom menggunakan `value_counts()` untuk melihat distribusi data setelah proses pembersihan.


# 3. Ganti missing value pada variabel numerik pakai mean

```python
# kolom numerik yang perlu diimputasi
kolom_numerik = ['length', 'width', 'height']

# pastikan tipenya numerik dulu
for col in kolom_numerik: # looping untuk setiap kolom numerik
    df[col] = pd.to_numeric(df[col], errors='coerce') # ubah ke tipe numerik, nilai tidak valid jadi NaN

# hitung mean tiap kolom terus isi yang kosong
for col in kolom_numerik: # looping lagi untuk proses imputasi
    rata2 = df[col].mean() # menghitung rata-rata tiap kolom
    print(f"\nMean {col}: {rata2:.2f}") # menampilkan nilai rata-rata
    df[col].fillna(rata2, inplace=True) # mengisi nilai kosong dengan rata-rata

print("\ncek missing value setelah diisi:")
print(df[kolom_numerik].isnull().sum())
```

Bagian ini digunakan untuk mengisi nilai kosong pada kolom numerik dalam dataset. Pertama, ditentukan kolom yang akan diproses yaitu `length`, `width`, dan `height`. Kemudian, setiap kolom diubah ke tipe numerik agar bisa dilakukan perhitungan, dan jika ada data yang tidak valid akan otomatis dianggap sebagai nilai kosong (NaN). Setelah itu, dihitung nilai rata-rata (mean) dari masing-masing kolom, lalu nilai yang kosong diisi menggunakan rata-rata tersebut. Terakhir, dilakukan pengecekan untuk memastikan bahwa tidak ada lagi nilai kosong yang tersisa pada kolom-kolom tersebut.


# 4. Min-Max Normalization pada wheel-bases

```python
# pastikan numerik
df['wheel-bases'] = pd.to_numeric(df['wheel-bases'], errors='coerce')

# rumus min max = (x - min) / (max - min)
min_wb = df['wheel-bases'].min() # mengambil nilai minimum dari kolom wheel-bases
max_wb = df['wheel-bases'].max() # mengambil nilai maksimum dari kolom wheel-bases

df['wheel-bases-normalized'] = (df['wheel-bases'] - min_wb) / (max_wb - min_wb)
# melakukan normalisasi min-max agar nilai berada di rentang 0–1 dan menyimpannya ke kolom baru

print("\nHasil min-max normalization wheel-bases:")
print(df[['wheel-bases', 'wheel-bases-normalized']].head(10))
```

Setelah itu untuk melakukan normalisasi data pada kolom wheel-bases menggunakan metode min-max normalization. Pertama, data pada kolom tersebut diubah ke tipe numerik agar bisa dihitung, dan jika ada nilai yang tidak valid akan dianggap sebagai NaN. Kemudian, diambil nilai minimum dan maksimum dari kolom tersebut. Setelah itu, setiap nilai diubah menggunakan rumus (x - min) / (max - min) sehingga hasilnya berada pada rentang 0 sampai 1. Nilai yang sudah dinormalisasi disimpan pada kolom baru bernama wheel-bases-normalized. Terakhir, ditampilkan beberapa baris data untuk melihat hasil normalisasi yang telah dilakukan.

# 5. Z-Score Normalization pada length, width, height

```python
# rumus z-score = (x - mean) / std
kolom_zscore = ['length', 'width', 'height']

for col in kolom_zscore:  # looping untuk setiap kolom dalam daftar kolom_zscore
    mean = df[col].mean() # menghitung nilai rata-rata (mean) dari kolom
    std = df[col].std()  # menghitung standar deviasi (std) dari kolom
    df[col + '-zscore'] = (df[col] - mean) / std  # menghitung Z-score dan menyimpannya ke kolom baru
    print(f"\nZ-score {col} selesai, mean={mean:.2f}, std={std:.2f}") # menampilkan hasil proses tiap kolom

print("\nHasil z-score normalization:")
print(df[['length', 'length-zscore', 'width', 'width-zscore', 'height', 'height-zscore']].head())


# simpan hasil
df.to_csv('data_tugas_2_cleaned.csv', index=False)
print("\nselesai, file disimpan")

```

Bagian terakhir untuk melakukan **normalisasi Z-score** pada kolom `length`, `width`, dan `height`. Untuk setiap kolom, dihitung nilai rata-rata (mean) dan standar deviasi (std), kemudian setiap data diubah dengan rumus *(x - mean) / std* agar memiliki skala yang lebih seimbang. Hasilnya disimpan pada kolom baru dengan tambahan `-zscore`. Setelah itu, ditampilkan beberapa data untuk melihat hasil normalisasi. Terakhir, dataset yang sudah dibersihkan dan diproses disimpan ke file baru bernama `data_tugas_2_cleaned.csv`.
